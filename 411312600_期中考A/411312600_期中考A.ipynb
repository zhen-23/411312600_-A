````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
def init_db():
    conn = sqlite3.connect("activity_system.db")
    cursor = conn.cursor()
    try:
        # 每次執行先刪除舊表，確保乾淨重建
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS activities")

        # 建立活動表
        cursor.execute("""
            CREATE TABLE activities (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT UNIQUE NOT NULL,
                event_date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 建立報名紀錄表
        cursor.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT NOT NULL,
                activity_name TEXT NOT NULL,
                count INTEGER NOT NULL,
                status TEXT NOT NULL
            )
        """)

        # 放入初始資料
        initial_activities = [
            ("迎新晚會", "2026-11-20", 6),
            ("程式工作坊", "2026-11-25", 4),
            ("籃球友誼賽", "2026-11-28", 3)
        ]
        cursor.executemany("""
            INSERT INTO activities (name, event_date, capacity)
            VALUES (?, ?, ?)
        """, initial_activities)

        conn.commit()
        print("✅ 資料庫與初始資料建立成功！")
    finally:
        conn.close()

init_db()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
def show_tables():
    conn = sqlite3.connect("activity_system.db")
    try:
        print("--- 【activities 活動表】 ---")
        df_act = pd.read_sql_query("SELECT * FROM activities", conn)
        display(df_act)

        print("\n--- 【registrations 報名紀錄表】 ---")
        df_reg = pd.read_sql_query("SELECT * FROM registrations", conn)
        display(df_reg)
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
AI_CACHE = {}

def parse_intent(text):
    if text in AI_CACHE:
        return AI_CACHE[text]

    prompt = f"""
你是一個活動報名系統的語意解析助理。請從使用者的句子中萃取三個欄位：動作、活動、人數。
- 動作只能從以下選擇：【報名、取消、查詢、統計、未知】
- 活動只能從以下選擇：【迎新晚會、程式工作坊、籃球友誼賽、未知】
- 人數：請找出數字（例如「兩個」代表 2，「我和兩個朋友」代表 3）。如果沒提到人數，預設為 1。

請嚴格按照以下格式回覆，不要有多餘的字：
動作,活動,人數

使用者句子：{text}
"""

    raw_response = ""
    for attempt in range(3):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            raw_response = response.choices[0].message.content.strip()
            time.sleep(2)
            break
        except Exception as e:
            if "429" in str(e) and attempt < 2:
                print("⚠️ 遭遇速率限制 (429)，等待 30 秒後重試...")
                time.sleep(30)
            else:
                raw_response = "未知,未知,1"
                break

    # 安全防護：嚴格過濾 AI 回覆，不直接信任
    valid_actions = ["報名", "取消", "查詢", "統計"]
    valid_activities = ["迎新晚會", "程式工作坊", "籃球友誼賽"]

    action, activity, count_val = "未知", "未知", 1

    parts = raw_response.split(",")
    if len(parts) >= 3:
        potential_action = parts[0].strip()
        potential_activity = parts[1].strip()
        potential_count_str = parts[2].strip()

        if potential_action in valid_actions:
            action = potential_action
        elif "統計" in text or "各活動" in text:
            action = "統計"

        if potential_activity in valid_activities:
            activity = potential_activity
        else:
            for act in valid_activities:
                if act in text:
                    activity = act
                    break

        # 用正規表示式抓出人數數字
        numbers = re.findall(r'\d+', potential_count_str)
        if numbers:
            count_val = int(numbers[0])
        else:
            if "兩" in text or "兩個" in text or "2" in text:
                count_val = 3
            elif "一個" in text or "1" in text:
                count_val = 1

    AI_CACHE[text] = (action, activity, count_val, raw_response)
    return action, activity, count_val, raw_response
````

````python
# ===========================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ===========================================================
test_sentences = [
    ("我想報名迎新晚會", "報名", "迎新晚會", 1),
    ("蔡純珍要取消程式工作坊", "取消", "程式工作坊", 1),
    ("籃球友誼賽剩多少名額", "查詢", "籃球友誼賽", 1),
    ("我和兩個朋友要報名迎新晚會", "報名", "迎新晚會", 3),
    ("我想報名神隱少女", "報名", "未知", 1),
    ("隨便說個話聊天", "未知", "未知", 1),
    ("各活動報名狀況", "查詢", "未知", 1),  # 修正這裡為查詢
    ("張三報名籃球友誼賽", "報名", "籃球友誼賽", 1)
]

passed_count = 0
print(f"=== AI 解讀測試開始（共 {len(test_sentences)} 句）===\n")

for i, (sent, exp_act, exp_item, exp_cnt) in enumerate(test_sentences, 1):
    act, item, cnt, raw = parse_intent(sent)
    is_pass = (act == exp_act and (item == exp_item or exp_item == "*"))
    if is_pass:
        passed_count += 1
    print(f"測試 {i}: \"{sent}\"")
    print(f"  -> AI 原始回覆: {raw}")
    print(f"  -> 系統解讀: 動作={act}, 活動={item}, 人數={cnt}")
    print(f"  -> 結果: {'✅ 通過' if is_pass else '❌ 失敗'}\n")

print(f"📊 測試結算：通過 {passed_count} / {len(test_sentences)} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
def process_user_request(student_name, text):
    action, activity, count, _ = parse_intent(text)

    # 先判斷是否為「統計」，因為統計不需要特定的活動名稱
    if action == "統計" or "統計" in text or "各活動" in text:
        return get_statistics()

    if activity == "未知" or action == "未知":
        return "❌ 無法辨識您的需求或活動名稱，請重新輸入（例如：我想報名迎新晚會）。"

    conn = sqlite3.connect("activity_system.db")
    cursor = conn.cursor()
    try:
        # 1. 檢查活動是否存在並取得名額資料
        cursor.execute("SELECT capacity FROM activities WHERE name = ?", (activity,))
        row = cursor.fetchone()
        if row is None:
            return f"❌ 找不到活動：{activity}"
        capacity = row[0]

        if action == "查詢":
            cursor.execute("""
                SELECT SUM(count) FROM registrations
                WHERE activity_name = ? AND status = '已報名'
            """, (activity,))
            reg_row = cursor.fetchone()
            registered_total = reg_row[0] if reg_row and reg_row[0] is not None else 0
            remaining = capacity - registered_total
            return f"📌 【{activity}】總名額：{capacity} 人，已報名：{registered_total} 人，剩餘名額：{remaining} 人。"

        elif action == "報名":
            if not student_name:
                return "❌ 報名必須填寫姓名！"
            if count < 1 or count > 3:
                return f"❌ 報名人數限制為 1 至 3 人（您輸入了 {count} 人），拒絕受理。"

            # 檢查是否重複報名同一個活動
            cursor.execute("""
                SELECT id FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity))
            if cursor.fetchone() is not None:
                return f"❌ {student_name} 已經報名過【{activity}】了，不可重複報名。"

            # 檢查個人同時報名活動數上限（最多 2 個活動）
            cursor.execute("""
                SELECT COUNT(DISTINCT activity_name) FROM registrations
                WHERE student_name = ? AND status = '已報名'
            """, (student_name,))
            active_act_count_row = cursor.fetchone()
            active_act_count = active_act_count_row[0] if active_act_count_row else 0
            if active_act_count >= 2:
                return f"❌ {student_name} 同時報名的活動已達上限（2個），無法再報名新活動。"

            # 檢查名額是否足夠
            cursor.execute("""
                SELECT SUM(count) FROM registrations
                WHERE activity_name = ? AND status = '已報名'
            """, (activity,))
            reg_row = cursor.fetchone()
            registered_total = reg_row[0] if reg_row and reg_row[0] is not None else 0

            if registered_total + count > capacity:
                return f"❌ 【{activity}】名額不足！目前剩餘 {capacity - registered_total} 人名額，您欲報名 {count} 人。"

            # 寫入資料庫
            cursor.execute("""
                INSERT INTO registrations (student_name, activity_name, count, status)
                VALUES (?, ?, ?, '已報名')
            """, (student_name, activity, count))
            conn.commit()
            return f"✅ 報名成功！{student_name} 成功報名【{activity}】共 {count} 人。"

        elif action == "取消":
            if not student_name:
                return "❌ 取消報名必須填寫姓名！"
            cursor.execute("""
                SELECT id FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity))
            reg_row = cursor.fetchone()
            if reg_row is None:
                return f"❌ 查無 {student_name} 在【{activity}】的有效報名紀錄，無法取消。"

            cursor.execute("""
                UPDATE registrations SET status = '已取消'
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity))
            conn.commit()
            return f"✅ 取消成功！{student_name} 已取消【{activity}】的報名，名額已釋出。"

    finally:
        conn.close()

def get_statistics():
    conn = sqlite3.connect("activity_system.db")
    try:
        query = """
            SELECT
                a.name AS 活動名稱,
                a.event_date AS 活動日期,
                a.capacity AS 總名額,
                COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.count ELSE 0 END), 0) AS 已報名人數,
                a.capacity - COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.count ELSE 0 END), 0) AS 剩餘名額
            FROM activities a
            LEFT JOIN registrations r ON a.name = r.activity_name
            GROUP BY a.name, a.event_date, a.capacity
        """
        df = pd.read_sql_query(query, conn)
        return df
    finally:
        conn.close()
````

````python
# ===========================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ===========================================================
workflow_tests = [
    ("蔡純珍", "我想報名迎新晚會"),
    ("蔡純珍", "查詢迎新晚會"),
    ("李小明", "我和兩個朋友要報名迎新晚會"),
    ("李小明", "我想報名程式工作坊"),
    ("李小明", "我想報名籃球友誼賽"), # 應該被拒絕（超過2個活動上限）
    ("李小明", "取消迎新晚會"),
    ("王大同", "報名迎新晚會"),
    ("系統管理員", "查詢各活動報名狀況")  # 修正這裡，加上查詢讓 AI 正確對應
]

print("=== 🚀 系統流程測試開始 ===\n")
for i, (name, text) in enumerate(workflow_tests, 1):
    print(f"步驟 {i} | 姓名: {name} | 說: \"{text}\"")
    result = process_user_request(name, text)
    if isinstance(result, pd.DataFrame):
        print("系統回應（統計表）：")
        display(result)
    else:
        print(f"系統回應: {result}")
    print("-" * 50)

print("\n📋 最終資料表狀態：")
show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 提醒：if IN_COLAB: demo.launch(share=True)
# ============================================================
def gradio_interface(student_name, user_text):
    if not user_text.strip():
        return "請輸入您的需求指令。"
    result = process_user_request(student_name.strip(), user_text.strip())
    if isinstance(result, pd.DataFrame):
        # 將 DataFrame 轉成字客串表格呈現於 Gradio
        return result.to_markdown(index=False)
    return result

demo = gr.Interface(
    fn=gradio_interface,
    inputs=[
        gr.Textbox(label="您的姓名（查詢或統計可不填）", placeholder="例如：蔡純珍"),
        gr.Textbox(label="您的需求指令", placeholder="例如：我想報名迎新晚會 / 各活動報名狀況")
    ],
    outputs=gr.Markdown(label="系統回應"),
    title="🎉 系學會活動報名系統",
    description="請輸入您的姓名與一句話需求，系統將自動為您完成報名、取消、查詢或統計。"
)

if __name__ == "__main__":
    if IN_COLAB:
        demo.launch(share=True)
    else:
        demo.launch()
````
